# Color Palette Reduction (Clustering-Based Image Compression)

Run the cells below in order. The original image should be saved as `original.png` in the same directory.
The quantized image will be saved as `reduced_palette.png`.

```python
import Extractor
image_path = "original.png"
output_path = "reduced_palette.png"
n_colors = 10
```

```python
try:
    clusters = Extractor.get_colors_numpy(image_path, n_colors=n_colors, max_iter=100)
    palette = Extractor.extract_palette(clusters)
    reduced_img = Extractor.apply_palette(image_path, palette)
    reduced_img.save(output_path)
except ValueError as e:
    print(e)
```
