Run the cells below in order. The original image should be saved as "original.png" in the same folder as "Driver.py". The result will be saved as "reduced_palette.png".

```python
from PIL import Image
import Extractor

clusters = Extractor.get_colors("original.png", 9)
```

```python
color_map={}
for cluster in clusters:
    new_color=tuple(map(int,cluster.center.coordinates))
    for p in cluster.points:
        color_map[tuple(p.coordinates)]=new_color

img=Image.open("original.png").convert("RGB")
img.thumbnail(size=[1000,16*1000//9])

pixels=img.load()

w,h=img.size

for x in range(w):
    for y in range(h):
        pixels[x,y]=color_map[pixels[x,y]]

img.save("reduced_palette.png")
```
